```python
from transformers import pipeline
# Question 1
# Define a long paragraph
long_paragraph = (
    "Artificial intelligence (AI) has become a transformative force across various industries. "
    "It encompasses machine learning, deep learning, natural language processing, and computer vision, among other fields. "
    "AI systems are designed to perform tasks that typically require human intelligence, such as problem-solving, learning, and decision-making. "
    "The rapid advancements in AI technologies are leading to significant changes in how businesses operate and how people interact with technology. "
    "Ethical considerations and responsible development are crucial as AI continues to evolve and become more integrated into society."
)

# Create a summarization pipeline
summarizer = pipeline("summarization")

# Call the summarization pipeline on the long paragraph
summary_result = summarizer(long_paragraph, min_length=30, max_length=100, do_sample=False)

# Extract the summary text
summary_text = summary_result[0]['summary_text']

# Print the original text length
print(f"Original text length (characters): {len(long_paragraph)}")
print(f"Original text length (words): {len(long_paragraph.split())}")

# Print the generated summary along with its length
print("\nGenerated Summary:")
print(summary_text)
print(f"Summary length (characters): {len(summary_text)}")
print(f"Summary length (words): {len(summary_text.split())}")
```

```python
from transformers import pipeline
# Question 2
# Create a text generation pipeline
generator = pipeline("text-generation", model="distilgpt2")

# Define the prompt
prompt = "In 2030, AI systems will"

# Call the text generation pipeline
# num_return_sequences=2 will generate two different continuations
# max_new_tokens=50 limits the length of the generated text
generated_texts = generator(
    prompt,
    num_return_sequences=2,
    max_new_tokens=50,
    do_sample=True, # Enable sampling for more diverse outputs
    temperature=0.7 # Control the randomness of the output
)

# Print the generated continuations
print(f"Prompt: {prompt}")

for i, generated_output in enumerate(generated_texts):
    print(f"Generated Continuation {i+1}:")
    full_text = generated_output['generated_text']
    # A simple way to show only the continuation, assuming prompt is at the start
    continuation = full_text[len(prompt):].strip()
    print(continuation)
```

```python
from transformers import pipeline
# Question 3
# Create a sentiment analysis pipeline
sentiment_analyzer = pipeline("sentiment-analysis")

# Define a list of five short movie reviews
movie_reviews = [
    "This movie was absolutely brilliant! I loved every moment.",
    "The plot was convoluted and the acting was subpar. A total waste of time.",
    "It was an okay film, neither good nor bad, just very average.",
    "A captivating storyline with stellar performances. Highly recommend!",
    "I fell asleep halfway through. Incredibly boring and predictable."
]

# Pass the list of reviews into the pipeline in one call
results = sentiment_analyzer(movie_reviews)

# Print each review together with its predicted label and confidence score
print("--- Sentiment Analysis Results ---")
for i, (review, result) in enumerate(zip(movie_reviews, results)):
    label = result['label']
    score = result['score']
    print(f"\nReview {i+1}: '{review}'")
    print(f"Sentiment: {label} (Confidence: {score:.2f})")
print("----------------------------------")
```

```python
from transformers import pipeline
# Question 1 alternate
def run_summarization():
    # 1. Initialize the summarization pipeline
    summarizer = pipeline("summarization", model="sshleifer/distilbart-cnn-12-6")

    # 2. Define a long paragraph (5 sentences)
    original_text = (
        "The exploration of Mars has been a focal point of modern space agencies for decades, "
        "driven by the desire to understand the planet's geological history and potential for life. "
        "Early missions like the Viking landers provided the first close-up images of the surface, "
        "revealing a desolate, rocky landscape that sparked intense scientific curiosity. "
        "In recent years, rovers like Curiosity and Perseverance have utilized advanced chemistry labs "
        "to analyze soil samples and search for organic compounds deep within the Martian crust. "
        "International collaborations have also increased, with various nations launching orbiters to "
        "map the atmosphere and search for liquid water beneath the polar ice caps. "
        "As technology advances, the ultimate goal remains a crewed mission to Mars, which would "
        "mark a historic milestone in human history and our expansion into the solar system."
    )

    # 3. Generate the summary
    summary_result = summarizer(
        original_text,
        max_length=50,
        min_length=20,
        do_sample=False
    )

    summary_text = summary_result[0]['summary_text']

    # 4. Calculate lengths
    orig_len = len(original_text.split())
    summ_len = len(summary_text.split())

    # 5. Output the results
    print("-" * 30)
    print(f"ORIGINAL TEXT (Length: {orig_len} words):")
    print(original_text)
    print("-" * 30)
    print(f"GENERATED SUMMARY (Length: {summ_len} words):")
    print(summary_text)
    print("-" * 30)
    print(f"Compression Ratio: {((orig_len - summ_len) / orig_len) * 100:.2f}% reduction")

if __name__ == "__main__":
    run_summarization()
```

```python
from transformers import pipeline
# Question 2 alternate
def generate_ai_futures():
    # Initialize the text-generation pipeline
    generator = pipeline("text-generation", model="gpt2")

    # Define the prompt
    prompt = "In 2030, AI systems will"

    # Call the pipeline
    results = generator(
        prompt,
        num_return_sequences=2,
        max_new_tokens=50,
        do_sample=True,
        temperature=0.8,  # Controls randomness (higher = more creative)
        truncation=True
    )

    # 4. Print and compare the results
    print(f"PROMPT: {prompt}\n" + "="*50)

    for i, output in enumerate(results):
        print(f"CONTINUATION {i+1}:")
        print(output['generated_text'].strip())
        print("-" * 50)

if __name__ == "__main__":
    generate_ai_futures()
```

```python
from transformers import pipeline
# Question 3 alternate
def analyze_movie_reviews():
    # 1. Initialize the sentiment-analysis pipeline
    # This loads a default model (DistilBERT fine-tuned on SST-2)
    classifier = pipeline("sentiment-analysis")

    # 2. Define a list of five movie reviews
    reviews = [
        "The cinematography was breathtaking and the acting was top-notch!",
        "A complete waste of time. The plot was non-existent and the pacing was glacial.",
        "It was an okay movie, but it didn't quite live up to the hype.",
        "I've never seen such a bold and original take on the sci-fi genre before.",
        "The ending was confusing and felt very rushed compared to the rest of the film."
    ]

    # 3. Pass the entire list into the pipeline in one call
    results = classifier(reviews)

    # 4. Print each review with its predicted label and confidence score
    print(f"{'REVIEW CONTENT':<60} | {'LABEL':<10} | {'CONFIDENCE'}")
    print("-" * 85)

    for review, result in zip(reviews, results):
        label = result['label']
        score = result['score']
        # Truncate review text for clean table formatting if necessary
        display_text = (review[:57] + '...') if len(review) > 60 else review

        print(f"{display_text:<60} | {label:<10} | {score:.4f}")

if __name__ == "__main__":
    analyze_movie_reviews()
```
